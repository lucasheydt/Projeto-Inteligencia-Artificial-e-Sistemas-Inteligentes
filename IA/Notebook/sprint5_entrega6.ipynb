```python
%load_ext autoreload
%autoreload 2

import sys
import torch
import os
from pathlib import Path

# Ajuste para a pasta dos seus scripts
sys.path.append(str(Path.cwd().parent / 'Codigo'))
sys.path.append(str(Path.cwd()))

from bpe import BPETokenizer
from model import GPTModel
from generation import generate_text_advanced

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f'Dispositivo: {device}')
```

```python
# A configuração DEVE ser idêntica à do treino!
CONTEXT_LENGTH = 256

bpe = BPETokenizer('gpt2')

MINI_GPT_CONFIG = {
    'vocab_size': bpe.vocab_size, 
    'context_length': CONTEXT_LENGTH, 
    'emb_dim': 256,   
    'n_heads': 8,     
    'n_layers': 6,    
    'drop_rate': 0.1, 
    'qkv_bias': False
}

# Instanciamos o cérebro 'vazio'
model = GPTModel(MINI_GPT_CONFIG).to(device)
print("Arquitetura Mini GPT (vazia) inicializada.")
```

```python
# INSIRA AQUI O NOME DO ARQUIVO GERADO NO SEU TREINAMENTO (ENTREGÁVEL 4)
ARQUIVO_PESOS = "pesos_exp_a_minigpt_lr_5e-4.pth"

if not os.path.exists(ARQUIVO_PESOS):
    print(f"[ERRO CRÍTICO] Arquivo '{ARQUIVO_PESOS}' não encontrado na pasta.")
    print("Por favor, verifique o nome do arquivo gerado no notebook do Entregável 4 e atualize a variável ARQUIVO_PESOS.")
else:
    # A função load_state_dict injeta as matrizes matemáticas de volta na rede
    model.load_state_dict(torch.load(ARQUIVO_PESOS, map_location=device))
    print(f"\n[SUCESSO] Pesos de '{ARQUIVO_PESOS}' carregados para o modelo.")
    
    # Colocamos o modelo em modo de avaliação (desativa camadas de treino como Dropout)
    model.eval()
    print("Modelo configurado para modo de inferência (eval).")
```

```python
texto_inicial = "Uma noite destas"
token_ids = bpe.encode(texto_inicial)
idx = torch.tensor(token_ids).unsqueeze(0).to(device)

print("--- INICIANDO INFERÊNCIA COM PESOS CARREGADOS ---")
print(f"Contexto Inicial: '{texto_inicial}'")

try:
    # Vamos usar uma temperatura ligeiramente controlada (0.8) para um bom balanço
    out_idx = generate_text_advanced(model, idx, max_new_tokens=40, context_size=CONTEXT_LENGTH, temperature=0.8)
    texto_gerado = bpe.decode(out_idx[0].tolist())
    print("\n--- TEXTO GERADO ---")
    print(texto_gerado)
except NameError as e:
    print("ERRO: A função 'generate_text_advanced' não foi importada. Execute o Entregável 5 para criar/atualizar o avaliacao.py.")
```
